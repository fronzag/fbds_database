# Exemplo de download de dados

## Instruções de uso

Segundo o site do [FBDS](https://geo.fbds.org.br/), atualmente os dados estão
organizados segundo a seguinte estrutura:

ESTADO > MUNICÍPIO > CATEGORIA

O downloader desenvolvido (`FBDSDownloader`) permite baixar:
- todos os dados disponíveis
- todas as categorias por um ou mais estados
- todos os estados por uma ou mais categorias
- por uma combinação entre estados e categorias


## Importação da classe

```python
from fbds.fbds import FBDSDownloader
```

## Criação de uma instância do downloader

```python
downloader = FBDSDownloader()
```

### Todos os dados disponíveis

```python
downloader.download(download_dir="./data")
```

### Todas as categorias por um ou mais estados

```python
# Todas as categorias em um estado
downloader.download(download_dir="./data", states=["AC"])

# Todas as categorias em mais de um estado
downloader.download(download_dir="./data", states=["PA", "AP"])
```

### Todos os estados por uma ou mais categorias

```python
# Uma categoria em todos os estados
downloader.download(download_dir="./data", categories=["USO"])

# Mais de uma categoria em todos os estados
downloader.download(download_dir="./data", categories=["USO", "HIDROGRAFIA"])
```

### Combinação entre estados e categorias

```python
# Combinação entre estados e categorias
downloader.download(
    download_dir="./data",
    states=["AC", "RR"],
    categories=["APP", "HIDROGRAFIA"],
)
```
